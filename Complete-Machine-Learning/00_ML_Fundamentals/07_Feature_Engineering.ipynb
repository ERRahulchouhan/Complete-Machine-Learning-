## Mathematical intuition
A feature map φ(x) can expose useful structure to a learning algorithm.

Always define the population, sample, and assumptions before interpreting a statistic.

```python
import numpy as np
import pandas as pd
rng = np.random.default_rng(42)
x = rng.normal(loc=5, scale=2, size=1000)
print({"n": len(x), "mean": float(x.mean()), "std": float(x.std(ddof=1))})
display(pd.Series(x, name="sample").describe())
```
